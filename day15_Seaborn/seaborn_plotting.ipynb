# Heatmap 

```python
# plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True)# axes level function
```

```python
temp_df  = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5,cmap='jet')# axes level function
```

# Multiplots

```python
# figure level function -> relplot -> displot -> catplot -> lmplot
# facet column 
# facet grid 
g=sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.violinplot,'sex','total_bill')
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.boxplot,'sex','total_bill')
```

# pair grid

```python
g=sns.PairGrid(data=tips,hue='sex')
g.map(sns.scatterplot)
```

```python
g=sns.PairGrid(data=tips)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```
